p.27

```python
class FourCal:
  def setdata(self, firs,second):
    self.firs=firs
    self.second = second
  def sum(self):
    result = self.firs + self.second
    return result
  def mul(self):
    result = self.firs * self.second
    return result
  def sub(self):
    result = self.firs - self.second
    return result
  def div(self):
    result = self.firs / self.second
    return result
a=FourCal()
a.setdata(4,2)
print(a.sum())
print(a.mul())
print(a.sub())
print(a.div())
```

```python
import random

player1_dice = []
player2_dice = []
for i in range(3):
  player1_dice.append(random.randint(1,6))
  player2_dice.append(random.randint(1,6))
print("Player 1 rolled " + str(player1_dice))
print("Player 2 rolled " + str(player2_dice))

if sum(player1_dice) == sum(player2_dice):
  print("Draw")
elif sum(player1_dice)> sum(player2_dice):
  print("Player 1 wins")
else:
  print("Player 2 wins")
```

```python
from random import randint

class Player:
  def __init__(self):
    self.dice = []

  def roll(self):
    self.dice = []
    for i in range(3):
      self.dice.append(randint(1,6))
  def get_dice(self):
    return self.dice

player1 = Player()
player2 = Player()

player1_dice = []
player2_dice = []

player1.roll()
player2.roll()

print("Player 1 rolled " + str(player1.get_dice()))
print("Player 2 rolled " + str(player2.get_dice()))


if sum(player1_dice) == sum(player2_dice):
  print("Draw")
elif sum(player1_dice)> sum(player2_dice):
  print("Player 1 wins")
else:
  print("Player 2 wins")
```
