### Increasing the `timeout` for `urlopen`

If the website is just slow to respond, you can increase the timeout duration. The default timeout can vary, but explicitly setting it can help in some cases. Let's try setting it to 10 seconds.

```python
from urllib.request import urlopen
from bs4 import BeautifulSoup

# Try to open the URL with a 10-second timeout
try:
    html = urlopen("http://pythonscraping.com/pages/page1.html", timeout=10)
    bsObj = BeautifulSoup(html.read(), "html.parser")
    print(bsObj.h1)
except Exception as e:
    print(f"An error occurred: {e}")
```
